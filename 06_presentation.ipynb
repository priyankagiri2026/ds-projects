# Data Cleaning

Looking at the data we realized obvious issues already.
- Changed data type 'str' for the 'date' column to 'datetime64[us]'
- Checked for rows with empty cells

Looking into specific colums even more issues showed up
- Column 'waterfront' was of type 'float64' while basically holding bolean values, so changed it to 'int64'
- Column 'yr_built' was of type 'float64' with values like '', which we could just convert to 'int64'
- Column 'yr_renovated' was a bit more tricky
   - type 'float64' but with values like '19920.', '20110.', but showing as pattern
   - so we converted it to 'str' deleted '0.' with replace and converted it to 'int64'

Finally we decided, that a complete inspection of the data will take to long for our project scope.
