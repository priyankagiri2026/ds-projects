# The Client

Larry Sanders wants to buy a house.

Larry is interested in a house that

- has a view to the waterfront
- is nice & isolated but central
- neighborhood without kids (while he has kids on his own)
- fits a limited budget

We don't know what "a limited budget" means. So we need to come back to the client, to learn about his limit.
At the same time, we will look for houses, so we're able to show prices and ranges for King County. 

# Data from KCHPP

Kings County Housing Price Project

The goal of this project was to create a linear regression model that would accurately predict housing prices in Kings County, Washington.
We fetched the data and went through some data cleaning before we started working on hypothesis related to our customer.
